```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
```

```python
df['readmitted'].value_counts()
```

```python
df['readmitted_30'] = (df['readmitted'] == '<30').astype(int)
```

```python
df['readmitted_30'].value_counts()
```

```python
features = [
    'race',
    'gender',
    'age',
    'admission_type_id',
    'time_in_hospital',
    'num_lab_procedures',
    'num_procedures',
    'num_medications',
    'number_outpatient',
    'number_emergency',
    'number_inpatient',
    'number_diagnoses',
    'max_glu_serum',
    'A1Cresult',
    'insulin',
    'change',
    'diabetesMed'
]

X = df[features]
```

```python
y = df['readmitted_30']
```

```python
X.head()
```

```python
y.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
categorical_columns = [
    'race',
    'gender',
    'age',
    'max_glu_serum',
    'A1Cresult',
    'insulin',
    'change',
    'diabetesMed'
]
```

```python
numerical_columns = [
    'admission_type_id',
    'time_in_hospital',
    'num_lab_procedures',
    'num_procedures',
    'num_medications',
    'number_outpatient',
    'number_emergency',
    'number_inpatient',
    'number_diagnoses'
]
```

```python
X_encoded = pd.get_dummies(
    X,
    columns=categorical_columns,
    drop_first=True
)
```

```python
X_encoded.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight='balanced'
)
```

```python
rf_model.fit(X_train, y_train)
```

```python
y_pred = rf_model.predict(X_test)
```

```python
y_pred[:10]
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
```

```python
print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1-score :", f1)
```

```python
pd.Series(y_pred).value_counts()
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
y_prob = rf_model.predict_proba(X_test)[:, 1]
```

```python
y_prob[:10]
```

```python
thresholds = [0.50, 0.40, 0.30, 0.20, 0.10, 0.05]

for threshold in thresholds:
    y_pred_threshold = (y_prob >= threshold).astype(int)
    
    recall = recall_score(y_test, y_pred_threshold)
    precision = precision_score(y_test, y_pred_threshold, zero_division=0)
    
    print("Threshold:", threshold)
    print("Recall:", recall)
    print("Precision:", precision)
    print()
```

```python
for threshold in thresholds:
    y_pred_threshold = (y_prob >= threshold).astype(int)

    recall = recall_score(y_test, y_pred_threshold)
    precision = precision_score(y_test, y_pred_threshold, zero_division=0)
    f1 = f1_score(y_test, y_pred_threshold, zero_division=0)

    print("Threshold:", threshold)
    print("Recall:", recall)
    print("Precision:", precision)
    print("F1-score:", f1)
    print()
```

```python
final_threshold = 0.10

y_pred_final = (y_prob >= final_threshold).astype(int)

print("Final Threshold:", final_threshold)
```

```python
from sklearn.metrics import confusion_matrix

final_cm = confusion_matrix(y_test, y_pred_final)

print(final_cm)
```

```python
#"As the classification threshold decreases, recall increases, while precision decreases. A threshold of 0.10 gives the highest F1-score among the tested thresholds, while 0.05 gives the highest recall."
```
